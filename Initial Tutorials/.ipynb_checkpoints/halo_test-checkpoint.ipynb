## Difference between DM Halo and DM particle data:
Dark Matter particles are simulated mass elements. 
Each particle has:
1. A position (x,y,z)
2. A velocity (vx, vy, vz)
3. A fixed mass (all particles have a mass equal to mp = 8.1×10⁷ h⁻¹ M☉ in fiducial cosmology)
4. An ID

Fig 3. in the paper basically shows how the dark matter particles clump together.

Dark Matter Halos are gravitationally bound clumps. Regions where particles have collapsed together. Each halo entry has:
1. Position
2. Mass (mvir, m200b, etc - sum of particle masses within some radius)
3. Virial ratio, spin, shape, etc

Physically, the represent the sites where galaxies would form. A halo is the dark matter "home" that a galaxy sits inside.

# Plotting the overdensity
At redshift 0, we can plot the positions of the dm particles (pos) to see how they clump (overdensity)

Some definitions:
grid: This is the grid resolution. It refers to how finely we chop up continuous space into discrete cells before counting particles in each cell. So each cell in our plot would be Lbox.grid size. ~= $195 h^-1$ kpc per cell side. Higher grid resolution -> smaller cells -> more noise if each cell isnt filled with enough particles. Lower grid resolution (smaller grid number) → bigger cells → smoother, blurrier image, but each cell averages over more particles so it's statistically more robust per cell (less shot noise), at the cost of washing out small-scale structure.


## Using NGP

```python
Lbox = sr.Lbox
grid = 512 #paper uses 1024 but since we have only 9% of particles we can decrease resolution
slab_thickness = Lbox/grid
z_center = Lbox/2 # slice through box center (pick any value in [0, Lbox))

# Select a thin slab along z:
mask = np.abs(pos[:, 2] - z_center) < slab_thickness/2 #Our z-slice is a very thin slice

x_slab = pos[mask, 0]
y_slab = pos[mask, 1]
print("Particles in slab:", x_slab.size)
```

```python
# Bin onto 2D grid:
density, xedges, yedges = np.histogram2d(x_slab, y_slab, bins=grid, range=[[0, Lbox], [0, Lbox]])

#Normalize by mean:
one_plus_delta = density/density.mean()

# Smooth using a Gaussian Kernel of radius ~185h^-1 kpc
smoothing_kpc = 185.0
sigma_pix = (smoothing_kpc/1000.0) / (Lbox/grid) #expressing the smoothing radius (in Mpc) in terms of slab thickness
one_plus_delta_smooth = gaussian_filter(one_plus_delta, sigma = sigma_pix)
```

## Using CIC:
CIC - Cloud in Cell. It's a particle to grid interpolation scheme, and this is more relevant that histogram2d() since this was used in the paper.

CIC treats each particle as a "Cloud" rather than a point and its "mass" gets split proportionally across the 2(in ID), 4(in 2D) or 8(in 3D) nearest cells, weighed by how much the particle's cloud overlaps each neighboring cell.

```python
import ctypes
```

```python
so_path = '/mnt/home/project/csatsangi.chetan/projects/sahyadri-sandbox/scripts/post-process/cic_update.so'
cic_lib = ctypes.CDLL(so_path)
```

```python
# Tell ctypes exactly what types each argument expects:
cic_lib.update_density.argtypes = [
    ctypes.c_long, # ndata
    ctypes.c_int, # grid
    np.ctypeslib.ndpointer(dtype=np.float64, ndim=1, flags='C_CONTIGUOUS'), # density
    np.ctypeslib.ndpointer(dtype=np.float64, ndim=1, flags='C_CONTIGUOUS'), # x
    np.ctypeslib.ndpointer(dtype=np.float64, ndim=1, flags='C_CONTIGUOUS'), # y
    np.ctypeslib.ndpointer(dtype=np.float64, ndim=1, flags='C_CONTIGUOUS'), # z
    ctypes.c_double, # cell_size
    ctypes.c_double, # interlace
]
cic_lib.update_density.restype = None
```

```python
Lbox = sr.Lbox
grid = 512

x = np.ascontiguousarray(pos[:,0], dtype=np.float64)
y = np.ascontiguousarray(pos[:,1], dtype=np.float64)
z = np.ascontiguousarray(pos[:,2], dtype=np.float64)
```

# Halo Occupancy Distribution Function


## HOD functions:

```python
#///////////////////////////////////////////////////////////////////
# HOD Fits
#///////////////////////////////////////////////////////////////////
class HODFits(object):
    """ Fitting functions and derivatives specifying best fit 5-parameter HOD fits 
         from Paul, Pahwa, Paranjape 2019, Table 3.
    """
    def __init__(self):
        """ Best fit HOD parameters and derivatives as function of luminosity threshold Mr. """

        # dictionary for best fit values of a0,a1,a2 for 5 HOD params
        # lgMmin,siglgM,lgM1,lgM0,alpha
        # fits are of the form a0 + a1*x + a2*x**2 for all except siglgM
        # in which case fit is a0 + a1*erf(x/a2)

        self.bf = {'lgMmin':np.array([12.33,-0.85,0.19]),
                   'siglgM':np.array([0.44,-0.16,0.3]),
                   'lgM1':np.array([13.52,-0.72,0.16]),
                   'lgM0':np.array([12.24,-0.54,0.0]),
                   'alpha':np.array([1.16,-0.20,0.10]), 
                   'prs':np.array([0.773, -0.065, 0.008])} #Red Fraction - Chetan

    def hod_fit(self,param,Mr):
        """ Return HOD parameter param (string) at luminosity threshold Mr.
             param can be one of 'lgMmin','siglgM','lgM1','lgM0','alpha', 'prs'.
        """
        a0,a1,a2 = self.bf[param]
        x = Mr + 20.5
        out = (a0 + a1*x + a2*x**2
               if param != 'siglgM' else
               a0 + a1*sysp.erf(x/a2))
        return out

    def hod_fit_deriv(self,param,Mr):
        """ Return derivative wrt Mr for HOD parameter param (string) at luminosity threshold Mr.
             param can be one of 'lgMmin','siglgM','lgM1','lgM0','alpha'.
        """
        a0,a1,a2 = self.bf[param]
        x = Mr + 20.5
        out = (a1 + 2*a2*x
               if param != 'siglgM' else
               (a1/a2)*2/np.sqrt(np.pi)*np.exp(-(x/a2)**2))
        return out

    

#///////////////////////////////////////////////////////////////////
# HOD Functions
#///////////////////////////////////////////////////////////////////
class HODFunctions(HODFits):
    """ Functions and derivatives specifying 5-parameter HODs 
         Uses fits from Paul, Pahwa, Paranjape 2019.
    """
    def __init__(self):
        """ HOD functions and derivatives as function of luminosity threshold Mr. """

        HODFits.__init__(self)

    def erf_arg(self,Mr,lgm):
        """ Convenience function.
             If both arguments are arrays, output has shape (lgm.size,Mr.size).
        """
        Mr_sc = np.isscalar(Mr)
        lgm_sc = np.isscalar(lgm)

        lgm_arr = np.outer(lgm,np.ones(Mr.size)) if ((not Mr_sc) & (not lgm_sc)) else 1.0*lgm
        erfarg = (lgm_arr - self.hod_fit('lgMmin',Mr))/self.hod_fit('siglgM',Mr)
        return erfarg

    def fcen_thresh(self,Mr,lgm):
        """ fcen(>L|m) = (1/2)[1 + erf(log10(m/Mmin)/siglgM)]. 
             If both arguments are arrays, output has shape (lgm.size,Mr.size).
        """
        erfarg = self.erf_arg(Mr,lgm)
        out = 0.5*(1+sysp.erf(erfarg))
        return out

    def Nsat_thresh(self,Mr,lgm):
        """ Nsat(>L|m) = ((m - M0)/M1)^alpha.
             If both arguments are arrays, output has shape (lgm.size,Mr.size).
        """
        M0 = 10**self.hod_fit('lgM0',Mr)
        M1 = 10**self.hod_fit('lgM1',Mr)
        alpha = self.hod_fit('alpha',Mr)
        mhalo = 10**lgm

        Mr_sc = np.isscalar(Mr)
        lgm_sc = np.isscalar(lgm)

        if Mr_sc & lgm_sc:
            out = ((mhalo - M0)/M1)**alpha if mhalo > M0 else 0.0
        else:
            mhalo_arr = np.outer(mhalo,np.ones(Mr.size)) if ((not Mr_sc) & (not lgm_sc)) else 1.0*mhalo
            Dm = (mhalo_arr - M0)/M1
            Dm[Dm < 0.0] = 0.0
            out = Dm**alpha

        return out
    
    
```

## Mocker Class:

```python
class Mocker(HODFunctions):
    def __init__(self, real, snap, sim_stem):
        self.real = real
        self.snap = snap
        self.sim_stem = sim_stem
        
        print("Initializing SnapshotReader...")
        sr = SnapshotReader(sim_stem=self.sim_stem, real=self.real, snap=self.snap, use_compressed=True)
        
        HODFunctions.__init__(self)
        self.redshift = sr.redshift
        self.Om = sr.Om
        self.Lbox = sr.Lbox # Mpc/h
        self.massdef = 'm200b'
        self.mmin = 2e11 # Msun/h
        self.Mrmax = -19 #Maximum(faintest) r-band absolute magnitude for galaxies
        self.Dvir = 200*self.Om #Virial density threshold
        
        self.rhoc = 2.7754e11 # (Msun/h) / (Mpc/h)^3 - Critical Density pf the universe at z=0
        self.rhoc_z = self.rhoc*self.EHub(self.redshift)**2 #Critical density at a particular redshift
        
        self.rng = np.random.RandomState(seed=42) #Random number generator

        dMr = 0.001 #Bin spacing for Mrvals
        nMr = int((23.5+self.Mrmax)/dMr) #number of Mrvals
        self.Mrvals = np.linspace(-23.5,self.Mrmax,nMr) # useful for interpolation later
        self.dMr = self.Mrvals[1]-self.Mrvals[0]
        
        # self.halodatatype = {'ID':int,'descID':int,
        #                     'mbnd_vir':float,'vmax':float,'vrms':float,'rvir':float,'rs':float,'np':int,
        #                     'x':float,'y':float,'z':float,'vx':float,'vy':float,'vz':float,
        #                     'Jx':float,'Jy':float,'Jz':float,'spin':float,'rs_klypin':float,
        #                     'mvir':float,'m200b':float,'m200c':float,'mCustom2':float,'mCustom':float,
        #                     'Xoff':float,'Voff':float,'spin_bullock':float,'b_to_a':float,'c_to_a':float,
        #                     'Ax':float,'Ay':float,'Az':float,'b_to_a_500c':float,'c_to_a_500c':float,
        #                     'Ax_500c':float,'Ay_500c':float,'Az_500c':float,
        #                     'TbyU':float,'Mpe_Behroozi':float,'Mpe_Diemer':float,'halfmassradius':float,
        #                     'pid':int} 

        
    ############################################################
    def EHub(self,z):
        """ Computes the dimensionless Hubble Parameter and a given redshift z."""
        return np.sqrt(self.Om*(1+z)**3 + 1-self.Om)
    ############################################################
    
    ############################################################
    # def read_this(self):
    #     halos = pd.read_csv(self.halo_cat,dtype=self.halodatatype,names=self.halodatatype.keys(),
    #                         comment='#',sep=r'\s+',header=None).to_records()
    #     return halos
    ############################################################
    

    ############################################################
    def prep_halo_data(self, massdef='m200b', QE=0.5,Npmin=0, sorthalos=False, keep_subhalos=False):
        """ Reads and cleans catalog by selecting parent objects with mass at least mmin.
            Returns structured array for full halo properties.
        """ 
        hr = HaloReader(sim_stem= self.sim_stem, real=self.real, snap=self.snap)
        hpos, halos = hr.prep_halos(massdef=massdef, QE=QE, Npmin=Npmin, sorthalos=sorthalos, keep_subhalos=keep_subhalos)
        
#         halos = self.read_this()
#         Nhalos_all = halos.size
#         cond_clean = (halos[self.massdef] >= self.mmin)
#         cond_clean = cond_clean & (halos['pid'] == -1)

#         halos = halos[cond_clean]
#         print("Kept {0:d} of {1:d} objects in catalog".format(halos.size,Nhalos_all))

#         del cond_clean
#         gc.collect()

        return hpos, halos
    ############################################################


    ############################################################
    def occupy_halos(self,halo_mass):
        # We use the Monte Carlo Rejection Smpling technique to occupy halos
        """ Decide which halos are occupied. 
             Input is 1-d array of halo masses. Output is 1-d boolean array of same size.
        """
        print("Occupying halos... ")

        occupy = np.ones(halo_mass.size,dtype=bool)
        fcen_min = self.fcen_thresh(self.Mrmax,np.log10(halo_mass))
        u = self.rng.rand(halo_mass.size) #draws a uniform random number between 0.0 to 1.0 for every halo(dice)
        #For any given halo, if the randum number u is greater than the calculated probability, it is unoccupied
        occupy[u > fcen_min] = False 

        print_string = "... {0:d} of {1:d} halos occupied\n".format(np.where(occupy)[0].size,occupy.size)
        print_string += "--------------------------------"
        print(print_string)

        return occupy
    ############################################################
    

    ############################################################
    def assign_centrals(self,halos):
        """ Allocate memory and assign (some) central properties. """

        print("Assigning values for centrals... ")
        ####################################################
        # Allocate memory for centrals
        centrals = np.zeros(halos.size,dtype=[('haloid','int'),
                                              ('lgm','double'),('rvir','double'),('con','double'),
                                              ('x','double'),('y','double'),('z','double'),
                                              ('Mr','double'),('Nsat','int'),('color', 'U4')])

        print("... halo IDs, halo masses, positions")
        centrals['haloid'] = halos['ID']
        centrals['lgm'] = np.log10(halos[self.massdef])
        centrals['x'] = halos['x']
        centrals['y'] = halos['y']
        centrals['z'] = halos['z']
        
        # ##Chetan Plays:
        # centrals['vx'] = halos['vx']
        # centrals['vy'] = halos['vy']
        # centrals['vz'] = halos['vz']
        
        rvir = (3*halos[self.massdef]/(4*np.pi*self.Dvir*self.rhoc_z))**(1/3.)
        # Units Mpc/h [physical]
        centrals['rvir'] = rvir*(1+self.redshift) # store comoving value
        centrals['con'] = rvir/(halos['rs']*1e-3/(1+self.redshift)) # rs is comoving kpc/h
        
        ####################################################
        # Assign central luminosity based on mass of halo
        centrals['Mr'] = self.assign_central_luminosity(centrals['lgm'])

        ####################################################
        # Poisson numbers for Nsat: NUmber of satellites for each central
        centrals['Nsat'] = self.get_Poisson(centrals['lgm'])

        print_string = '... done\n'
        print_string += "--------------------------------"
        print(print_string)

        return centrals
    ############################################################
        


    ############################################################
    def get_Poisson(self,lgm_halos):
        """ Calculate mean Nsat and generate Poisson numbers of satellites. 
            Can be generalised to depend on more than halo mass.
        """
        print("... Poisson satellite counts")
        mean_Nsat = self.Nsat_thresh(self.Mrmax,lgm_halos)
        Nsat_values = self.rng.poisson(mean_Nsat)

        return Nsat_values
    ############################################################
    

    ############################################################
    def assign_central_luminosity(self,lgm_halos):
        """ Assign central luminosities. Can be generalised to depend on more than halo mass. """
        
        print("... luminosities")
        Mr_cen = np.zeros(lgm_halos.size,dtype=float)

        ############################################
        # Generate uniform deviates
        u = self.rng.rand(lgm_halos.size)

        ################################################
        for h in range(lgm_halos.size):
            # Luminosity function is Pcen(>L|m) = fcen(>L|m)/fcen(>Lmin|m)
            PcenL = self.fcen_thresh(self.Mrvals,lgm_halos[h])/self.fcen_thresh(self.Mrmax,lgm_halos[h])
            # keeping this in loop is not much slower than vectorised calc, but more memory-efficient

            #finding the bound of u :
            # Find largest L (smallest Mr) where Pcen(>L|m) > u
            Mr_upp = self.Mrvals[PcenL > u[h]][0]

            # Find smallest L (largest Mr) where Pcen(>L|m) <= u
            Mr_low = self.Mrvals[PcenL <= u[h]][-1]
            # choose fainter of the two (enforce nesting of HOD) : A halo capable of holding a bright galaxy must dynamically be capable of holding a fainter one
            Mr_cen[h] = np.max([Mr_upp,Mr_low])
        ################################################
        
        del u
        gc.collect()

        return Mr_cen
    ############################################################

    ########################################################
    def assign_centrals_colors(self, centrals, satellites, Mr_bin_edges = [-22, -21, -20, -19]):
        color = np.empty(centrals.size, dtype ='U4')
        u = self.rng.rand(centrals.size)
        
        #From Zehavi et al(SDSS)
        p_red_bin = {
            (-22, -21): 0.63,
            (-21, -20): 0.53,
            (-20, -19): 0.43,
        }
        
        for i in range(len(Mr_bin_edges)-1):
            Mr_hi, Mr_lo = Mr_bin_edges[i], Mr_bin_edges[i+1]
            bin_key = (Mr_hi, Mr_lo)
            p_red = p_red_bin.get(bin_key)
            
            #Fractions:
            in_bin_cen = (centrals['Mr'] >= Mr_hi) & (centrals['Mr'] < Mr_lo)
            in_bin_sat = (satellites['Mr'] >= Mr_hi) & (satellites['Mr'] < Mr_lo)
            n_cen, n_sat = in_bin_cen.sum(), in_bin_sat.sum()
            p_cen_bar = n_cen/(n_cen+n_sat)
            p_sat_bar = n_sat/(n_cen+n_sat)
            
            #prs:
            prs_per_gal = np.clip(self.hod_fit('prs', centrals['Mr'][in_bin_cen]), 0.01, 0.99)
            
            #Eq A.10 from paul,pahwa,paranjape(2019):
            p_red_cen  =(p_red - prs_per_gal*p_sat_bar)/p_cen_bar
            is_red = u[in_bin_cen] < p_red_cen
            color[in_bin_cen] = np.where(is_red, 'red', 'blue')
            
        return color

    ########################################################

    

    ########################################################
    def assign_satellites(self,centrals):
        """ Allocate memory and assign (some) satellite values."""
        print("Assigning values for satellites... ")

        Nsat_tot = np.sum(centrals['Nsat']) #Total Number of satellites to be assigned

        ####################################################
        # Allocate memory for satellites
        satellites = np.zeros(Nsat_tot,dtype=[('haloid','int'),
                                              ('lgm','double'),('con','double'),
                                              ('x','double'),('y','double'),('z','double'),
                                              ('Mr','double'), ('color', 'U4')])

        print("... halo properties")
        s_lo = 0
        for h in range(centrals.size):
            s_hi = s_lo + centrals['Nsat'][h]
            ################################################
            if centrals['Nsat'][h]:
                sl = np.s_[s_lo:s_hi]
                ############################################
                satellites['haloid'][sl] = centrals['haloid'][h]
                satellites['lgm'][sl] = centrals['lgm'][h]
                satellites['con'][sl] = centrals['con'][h]
            ################################################
            # Update
            s_lo = s_hi

        print("... positions")
        s_lo = 0
        for h in range(centrals.size):
            s_hi = s_lo + centrals['Nsat'][h]
            ################################################
            if centrals['Nsat'][h]:
                sl = np.s_[s_lo:s_hi]
                ############################################
                sx,sy,sz = self.assign_satellite_pos(centrals,h)
                satellites['x'][sl] = sx
                satellites['y'][sl] = sy
                satellites['z'][sl] = sz
            ################################################
            # Update
            s_lo = s_hi

        print("... luminosities")
        s_lo = 0
        for h in range(centrals.size):
            s_hi = s_lo + centrals['Nsat'][h]
            ################################################
            if centrals['Nsat'][h]:
                ############################################
                sMr = self.assign_satellite_luminosities(centrals,h)
                satellites['Mr'][s_lo:s_hi] = sMr
            ################################################
            # Update
            s_lo = s_hi
            
        print("... color")
        satellites['color'] = self.assign_satellite_colors(satellites)
        
        gc.collect
        

        print_string = '... done\n'
        print_string += "--------------------------------"
        print(print_string)

        return satellites
    ########################################################
    
    
    ########################################################
    def assign_satellite_luminosities(self,centrals,h):
        """ Assign satellite luminosities. """

        Nsat_h = centrals['Nsat'][h]
        Mr_cen = centrals['Mr'][h]
        lgmhalo = centrals['lgm'][h]
        PsatL = self.Nsat_thresh(self.Mrvals,lgmhalo)/self.Nsat_thresh(self.Mrmax,lgmhalo)

        ####################################################
        # Generate luminosities, without worrying about Lcen.
        # ... generate Nsat_h uniform deviates
        u = self.rng.rand(Nsat_h)
        # ... find largest L (smallest Mr) where Psat(>L|m) > u
        Mr_sat = np.array([self.Mrvals[(PsatL >= u[s])][0] for s in range(Nsat_h)])

        return Mr_sat
    ########################################################

    
    ########################################################
    def assign_satellite_colors(self, satellites):
        """ Red/blue assignment for satellites using prs(Mr), indipendent of halo mass"""
        
        prs = self.hod_fit('prs', satellites['Mr'])
        prs = np.clip(prs, 0.01, 0.99)
        u = self.rng.rand(satellites.size)
        is_red = u < prs
        color = np.where(is_red, 'red', 'blue')

        return color
    ########################################################

    ########################################################
    def assign_satellite_pos(self,centrals,h):
        """ Assign satellite positions and velocities acc to NFW profile using halo concentration. """
        # Number of satellites in this halo
        Nsat_h = centrals['Nsat'][h]
        # Virial radius [comoving Mpc/h]
        rvir = centrals['rvir'][h]
        con = centrals['con'][h]

        sat_pos = self.gen_NFW_profile(Nsat_h,cvir=con,Rvir=rvir,rng=self.rng)
        sx = sat_pos[:,0] + centrals['x'][h]
        sy = sat_pos[:,1] + centrals['y'][h]
        sz = sat_pos[:,2] + centrals['z'][h]

        sx = sx % self.Lbox
        sy = sy % self.Lbox
        sz = sz % self.Lbox

        return sx,sy,sz
    
    
    ############################################################
    def gen_NFW_profile(self,Nsat,cvir=None,Rvir=None,rng=None,include_cen=False,clean_up=False):
        """ Generate 3-d points distributed according to specified radial profile.
            Returns array of shape (Ngal,3) where Ngal = Nsat if include_cen = False else Nsat+1
            and central is assumed to be at origin. (Can include spatial offset later.)
        """
        if rng is None:
            rng = np.random.RandomState(42)
        phi = 2*np.pi*rng.rand(Nsat)
        cos_theta = 2*rng.rand(Nsat) - 1
        sin_theta = np.sqrt(1-cos_theta**2)
        
        rsamp = self.gen_rsamp(Nsat,cvir=cvir,Rvir=Rvir,rng=rng,clean_up=clean_up)

        x_trc = rsamp*sin_theta*np.sin(phi)
        y_trc = rsamp*sin_theta*np.cos(phi)
        z_trc = rsamp*cos_theta

        sample = np.array([x_trc,y_trc,z_trc]).T
        if include_cen:
            sample = np.append(sample,[[0.,0.,0.]],axis=0)
            
        if clean_up:
            del x_trc,y_trc,z_trc
            gc.collect()

        return sample
    ############################################################
    
    
    ############################################################
    def gen_rsamp(self,Nsat,cvir=None,Rvir=None,rng=None,clean_up=False):
        """ Generate radial location sample for speficied profile. 
            Return array of r values of shape (Nsat,).
        """
        xmax = 2*cvir

        if rng is None:
            rng = np.random.RandomState(42)

        xfine = np.linspace(0,xmax,100000)
        dx = xfine[1]-xfine[0]
        Px = np.zeros_like(xfine)
        
        rs = Rvir/cvir
        Px = self.Mencl_nfw(xfine)
        ind_small = np.where(xfine < 1e-3)[0]
        if ind_small.size:
            Px[ind_small] = xfine[ind_small]**2/2 - 2*xfine[ind_small]**3/3.0 
            Px[ind_small] = Px[ind_small] + 3*xfine[ind_small]**4/4.0 - 4*xfine[ind_small]**5/5.0
        Px /= self.Mencl_nfw(cvir)
                
        vran = rng.rand(Nsat)
        rsamp = np.interp(vran,Px,xfine)
        if np.any(rsamp <= dx):
            print('! Warning ! : Ignore separations x < {0:.2e}'.format(dx))

        if clean_up:
            del xfine,Px
            gc.collect()
        
        rsamp *= rs
        
        return rsamp
    ############################################################
    
    
    ############################################################
    def Mencl_nfw(self,x):
        """ Convenience function for NFW enclosed mass. Expect x = r/rs. """
        return np.log(1+x) - x/(1+x)
    ############################################################
    
    

    ############################################################
    def mock_it(self, massdef='m200b', QE=0.5,Npmin=0, sorthalos=False, keep_subhalos=False):
        """ Main execution routine. """        
        # halo data
        
        hr = HaloReader(sim_stem= self.sim_stem, real=self.real, snap=self.snap)
        print('HaloReader Initialized.')
        print('Reading halos...')
        hpos, halos_all = hr.prep_halos(massdef=massdef, QE=QE, Npmin=Npmin, sorthalos=sorthalos, keep_subhalos=keep_subhalos)
        # hpos, halos_all = self.prep_halo_data()

        # occupy halos
        occupy = self.occupy_halos(halos_all[self.massdef])
        halos = halos_all[occupy]
        all_masses = halos_all[self.massdef]
        del halos_all
        gc.collect()

        # galaxy properties except color and derived props
        centrals = self.assign_centrals(halos)
        satellites = self.assign_satellites(centrals)
        
        print("...Assigning centrals color")
        centrals['color'] = self.assign_centrals_colors(centrals=centrals, satellites=satellites)
        
        del halos
        gc.collect()

        print_string = "... {0:d} galaxies created\n".format(centrals.size + satellites.size)
        print_string += "--------------------------------"
        print(print_string)

        print_string = '... all done\n'
        print_string += "--------------------------------"
        print(print_string)

        return centrals,satellites,all_masses

    ############################################################
    
```

## 2pcf of galaxies:

### 2pcf:
self.proj: 
    If false, calculates the standard 3D spatial correlation function $\xi (r)$. 
    If true, it takes the 3D cf and integrates (projects) it along the line-of-sight to calculate the projected correlation function $w_p(r_p)$, which helps avoid redshift space distortions.


DD_split:
    Data-Data Pairs. DD_split uses a KD-tree to count how many pairs of galaxies are seperated by a given distance r. If pos_data2 is provided, it does a cross correlation(pairs between set 1 and set 2); otherwise it does an auto-correlation(pairs with set 1)
    
Projected $\xi$:
    Abel integral transformation:
    $$w_p(r_p) = 2 \int_{r_p}^{\sqrt{r_p^2 + \pi_{max}^2}} \frac{r \, \xi(r)}{\sqrt{r^2 - r_p^2}} dr$$
    Because the numerator goes to zero when $r$ goes to $r_p$, resulting in a singularity, the code handles this in two pieces:
    A. Interpolation for numerical smoothness: Takes the rough 3D function values cf_r and interpolates them onto a much finer, smoother grid (self.rmid_int ) so the calculus integration is accurate
    B. For every projected distance bin $r_p$ (self.rpmid[rp]), it filters the fine grid (cond) to only look at distances between $r_p$ and the maximum line-of-sight integration limit ($\sqrt{r_p^2 + \pi_{max}^2}$). Then, $Dr_p = r_int - r_p$ represents an extremely narro "step" right next to the boundary.
    C. For the singularity, the code uses a taylor series expansion of the integrand near the singularity boundary:
    $$\int_{r_p}^{r_p + \Delta r_p} \frac{r \, \xi(r)}{\sqrt{r^2 - r_p^2}} \, dr \approx \xi(r_p) \cdot \sqrt{2 \cdot r_p \cdot \Delta r_p}$$ And so, 
    $$\text{add\_this} = \sqrt{\frac{2 \cdot \Delta r_p}{r_p}} \cdot \xi(r_{int}) \cdot r_p$$
    D. The rest of the integral is found using the trapezoidal method using np.trap() using rmid_int[cond] as the points of integration
    

```python
class TwoPointCorrelationFunctionPeriodic(object):
    """ 2-point (cross-)correlation functions in bins of seperation.
        Assumes complete, cubic periodic box throughout
    """

    def __init__(self, lgbin=1, rmin=1e-2, rmax=3e1, nbin=15, Lbox=300.0, proj=0, pimax=60.0):
        """ Initialise the following:
            -- lgbin     : use logarithmic (1) or linear(0) binning
            -- rmin, rmax: min, max seperations in r(or rp)
            -- nbin      : number of bins in r(or rp)
            -- Lbox      : Box size in same units as rmin, smax, pimax.
            -- proj      : if 1, calculated projected correlation function wp(rp), else xi(r)
            --pimax      : max seperation in pi (bins in pi will be rp-deoendent)

            Methods:
            pair_counts, DD_only, RR_only, twoPCF
        """

        self.lgbin = lgbin
        self.rmin = rmin
        self.rmax = rmax
        self.nbin = nbin
        self.Lbox = Lbox
        self.proj = proj
        self.pimax = pimax

        self.TINY = 1e-15
        self.N_SPLIT = 100000 # max number of objects that can be handled on 1 cpu

        print_string = "Initialising 2pcf calculator for periodic boxes\n"
        print_string += "---------------------------"
        print(print_string)

        if self.proj:
            print("...Projected 2pcf calculated")
            if self.lgbin == 1:
                self.rpbin = np.logspace(np.log10(self.rmin), np.log10(self.rmax), self.nbin+1)
                self.rpmid = np.sqrt(self.rpbin[1:]*self.rpbin[:-1])

            else:
                self.rpbin = np.linspace(self.rmin, self.rmax, self.nbin+1)
                self.rpmid = 0.5*(self.rbin[1:]*self.rbin[:-1])
            self.NR = 50
            self.NR_INT = self.NR*10
            self.RMIN = self.rmin
            self.RMAX = 0.5*self.Lbox # NOTE THIS
            self.rbin = np.logspace(np.log10(self.RMIN),np.log10(self.RMAX),self.NR+1)
            self.rmid = np.sqrt(self.rbin[1:]*self.rbin[:-1])
            self.rbin_int = np.logspace(np.log10(self.RMIN),np.log10(self.RMAX),self.NR_INT+1)
            self.rmid_int = np.sqrt(self.rbin_int[1:]*self.rbin_int[:-1])
            self.dlnr_int = np.log(self.rmid_int[1]/self.rmid_int[0])

        else:
            print("...monopole 2pcf calculated")
            if self.lgbin == 1:
                self.rbin = np.logspace(np.log10(self.rmin),np.log10(self.rmax),self.nbin+1)
                self.rmid = np.sqrt(self.rbin[1:]*self.rbin[:-1])
            else:
                self.rbin = np.linspace(self.rmin,self.rmax,self.nbin+1)
                self.rmid = 0.5*(self.rbin[1:]+self.rbin[:-1])

        print_string = "...initialisation complete\n"
        print_string += "------------------------------"
        print(print_string)
        
    #############################################################


    #############################################################
    def pair_counts(self, pos_1, pos_2):
        """ Unnormalised binned pair counts
            between two data sets
            pos_j should have shape (ndata_j, 3) and
            Returns vector of length self.nbin.
        """
        tree_1 = syspat.cKDTree(pos_1, boxsize=self.Lbox)
        tree_2 = syspat.cKDTree(pos_2, boxsize=self.Lbox)

        cum_counts = tree_1.count_neighbors(tree_2, self.rbin, cumulative=True)
        bin_counts = np.diff(cum_counts)

        del tree_1, tree_2
        gc.collect()

        return bin_counts

    #############################################################

    #############################################################
    def RR_theory(self):
        print("... ...calculating RR from theory")
        #valid for rmax < Lbox/2, else see Deserno 04
        return 4*np.pi/3*(self.rbin[1:]**3-self.rbin[:-1]**3)/self.Lbox**3
    #############################################################

    #############################################################
    def twoPCF(self, pos_data, pos_data2=None):
        """ Auto/cross-correlation of data points.
            Assumes pos)data.shape() = (ndata, 3)
            Appliex recursive binary split for large data sets.
            Returns Peebles Hauser estimator DD/RR -1.
        """

        if pos_data.shape[1] != 3:
            raise TypeError("Incompatible data shape. Expected (ndata, 3).")

        DD = self.DD_split(pos_data, pos_data2=pos_data2)

        RR = self.RR_theory()
        cf_r = DD/(RR+self.TINY) - 1.0

        if self.proj:
            print("... ...projecting")
            cf = np.zeros(self.nbin, dtype=float)
            # Softening the Integrand:
            cf_r_int = np.interp(self.rmid_int, self.rmid, cf_r)
            
            # Calculating cf projected along the perpendicular direction for each rp:
            for rp in range (self.nbin):
                cond = (self.rmid_int > self.rpmid[rp]) & (self.rmid_int**2 < (self.rpmid[rp]**2 + self.pimax**2))
                ind  = np.where(cond)[0]
                if ind.size:
                    Drp = self.rmid_int[ind[0]] - self.rpmid[rp]
                    add_this = np.sqrt(2*Drp/self.rpmid[rp])*cf_r_int[ind[0]]*self.rpmid[rp]
                    cf[rp] = add_this + 2*np.trapz((self.rmid_int[cond]*cf_r_int[cond]
                                                    /np.sqrt(self.rmid_int[cond]**2 - self.rpmid[rp]**2)),
                                                   x=self.rmid_int[cond])

        else:
            cf = 1.0*cf_r

        print("... ... done")

        return cf
    #############################################################

    #############################################################
    def DD_only(self, pos_data1, pos_data2=None):
        """ DD (or D1D2) calculaiton
            Assumes pos_data.shape() = (ndata, 3)
            If pos_data2 is not None, assumes pos_data2.shape() = (ndata2,3).
            Returns DD or D1D2.
        """

        ndata1 = pos_data1.shape[0]
        
        if pos_data2 is None:
            cf = 1.0*self.pair_counts(pos_data1, pos_data1)/ndata1/(ndata1-1)
        else:
            ndata2 = pos_data2.shape[0]
            cf = 1.0*self.pair_counts(pos_data1, pos_data2)/ndata1/ndata2

        return cf
    #############################################################

    #############################################################

    def DD_split(self, pos_data, pos_data2=None):
        """ DD calculation wrapper.
            Assumes pos_data.shape() = (ndata, 3)
            If pos_data2 is not None, assumes pos_data2.shape() = (ndata2,3).
            Returns DD (or D1D2), applying binary recursive split for large data sets.
        """

        ndata = pos_data.shape[0]
        ndata_by_2 = ndata//2
        if ndata < self.N_SPLIT:
            print_string = '... ... calculating '
            print_string += "DD" if pos_data2 is None else "D1D2"
            print(print_string)
            cf = self.DD_only(pos_data, pos_data2=pos_data2)

        else:
            print("... ... binary split")
            if pos_data2 is None:
                pd1s = pos_data[:ndata_by_2].copy()
                ndata1s = pd1s.shape[0]
                pd1spr = pos_data[ndata_by_2:].copy()
                ndata1spr = pd1spr.shape[0]
                D1D1 = self.DD_split(pd1s)
                D1D1pr = self.DD_split(pd1s, pos_data2=pd1spr)
                D1prD1pr = self.DD_split(pd1spr)
                cf = ndata1s*(ndata1s-1)*D1D1 + ndata1s*ndata1spr*D1D1pr + ndata1spr*(ndata1spr-1)*D1prD1pr
                cf = 1.0*cf/ndata/(ndata-1)
                del pd1s, pd1spr
                gc.collect()

            else:
                ndata2 = pos_data2.shape[0]
                ndata2_by_2 = ndata//2
                pd1s = pos_data[:ndata_by_2].copy()
                ndata1s = pd1s.shape[0]
                pd1spr = pos_data[ndata_by_2:].copy()
                ndata1spr = pd1spr.shape[0]
                pd2s = pos_data2[:ndata2_by_2].copy()
                ndata2s = pd2s.shape[0]
                pd2spr = pos_data2[ndata2_by_2:].copy()
                ndata2spr = pd2spr.shape[0]
                D1D2 = self.DD_split(pd1s,pos_data2=pd2s)
                D1D2pr = self.DD_split(pd1s,pos_data2=pd2spr)
                D1prD2 = self.DD_split(pd1spr,pos_data2=pd2s)
                D1prD2pr = self.DD_split(pd1spr,pos_data2=pd2spr)
                cf = ndata1s*ndata2s*D1D2 + ndata1s*ndata2spr*D1D2pr 
                cf = cf + ndata1spr*ndata2s*D1prD2 + ndata1spr*ndata2spr*D1prD2pr
                cf = 1.0*cf/ndata/ndata2
                del pd1s,pd1spr,pd2s,pd2spr
                gc.collect()

        return cf
                
    #############################################################
        
                    
        
    
```

```python
sr = SnapshotReader(real=real, snap=snap, sim_stem=sim_stem, use_compressed=True)
tpcf_real = TwoPointCorrelationFunctionPeriodic(rmin=0.01, rmax=30, nbin=20, Lbox=sr.Lbox, proj=0)
tpcf = TwoPointCorrelationFunctionPeriodic(rmin=0.01, rmax=30, nbin=20, Lbox=sr.Lbox, proj=1, pimax=40.0)
```

## Colour Dependent Clustering:

1. g-r value:<br>
g and r are two photometric filters (green-ish and red-ish bands) - each galaxy has an apparent(or absolute) magnitude measured through each filter, M_g and M_R. <br>
$g-r = M_g - M_r$ <br>
Therefore a galaxy more brighter in red filter than green will have a more negative M_r and so g-r will be positive. This galaxy would therefore be "red". Similarly, a negative g-r value indicates a "blue" galaxy.

2. (g-r)cut:<br>
A single dividing line to sort every galaxy into a bucket - "red" or "blue" for clustering analysis. In Paul,Pahwa, Paranjape 2019:  
$(g-r)_{cut} = 0.21-0.03 \times M_r$ <br>
This is not a fixed threshold, it depends on the galaxy's own Mr value. Brighter galaxies have a more negative Mr get a _higher_ color cut. This reflects an observational fact: even within "Blue cloud" or "Red sequence" typically colour shifts slightly with luminosity, so a single fixed cut across all luminosities would misclassify galaxies at the bright/faint ends. So, for each galaxy: <br>
if (g-r) > (g-r)cut: galaxy is Red <br>
else: galaxy is blue

```python
#All reds:
c_red = (centrals['color'] == 'red')
cpos_red = cpos[c_red]
s_red = satellites['color'] == 'red'
spos_red = spos[s_red]
gpos_red = np.vstack((cpos_red, spos_red))
```

```python
#All blues:
cpos_blue = cpos[~c_red]
spos_blue = spos[~s_red]
gpos_blue = np.vstack((cpos_blue, spos_blue))
```

```python
from concurrent.futures import ProcessPoolExecutor

def run_2pcf(pos):
    return tpcf.twoPCF(pos, pos)

with ProcessPoolExecutor(max_workers=3) as executor:
    future_red = executor.submit(run_2pcf, gpos_red)
    future_blue = executor.submit(run_2pcf, gpos_blue)
    future_gal = executor.submit(run_2pcf, gpos)
    
    wp_red = future_red.result()
    wp_blue = future_blue.result()
    wp_gal = future_gal.result()
```

```python
plt.xscale('log')
plt.yscale('log')
plt.xlabel("$r_{{\\rm p}} \\, (h^{{-1}}{{\\rm Mpc}})$")
plt.ylabel("$w_{{\\rm p}}(r_{{\\rm p}})$")
plt.plot(tpcf.rpmid, wp_red, 'r-', label='red galaxies')
plt.plot(tpcf.rpmid, wp_blue, 'b-', label='blue galaxies')
plt.plot(tpcf.rpmid, wp_gal, 'k-', label='all galaxies')
plt.title('2-point corellation function')
plt.legend()
plt.savefig('2pcf_color.pdf', dpi=150)
plt.show()
```

## Using Corrfunc:

```python
from Corrfunc.theory import wp
```

```python
Lbox = mm.Lbox
pimax=60.0
nthreads = 4

rmin = 0.01
rmax = (1.0/5.0)*Lbox
nbin = 15
rp_bins = np.logspace(np.log10(rmin), np.log10(rmax), nbin+1)


cpos = np.array([centrals['x'], centrals['y'], centrals['z']]).T
spos = np.array([satellites['x'], satellites['y'], satellites['z']]).T
gpos = np.vstack((cpos, spos))
```

```python
def compute_wp(pos, rp_bins, Lbox, pimax=40, nthreads=4):
    x, y, z = np.ascontiguousarray(pos[:,0]), np.ascontiguousarray(pos[:,1]), np.ascontiguousarray(pos[:,2])
    results = wp(boxsize=Lbox, pimax=pimax, nthreads=nthreads, binfile=rp_bins, X=x, Y=y, Z=z, output_rpavg=True)
    return results
```

```python
wp_cen_result = compute_wp(cpos, rp_bins, Lbox, pimax, nthreads)
wp_sat_result = compute_wp(spos, rp_bins, Lbox, pimax, nthreads)
wp_gal_result = compute_wp(gpos, rp_bins, Lbox, pimax, nthreads)
```

```python
wp_cen_result.dtype.names
```

```python
fig, ax = plt.subplots(figsize=(7,7))

ax.set_xscale('log')
ax.set_yscale('log')
ax.plot(wp_gal_result['rpavg'], wp_gal_result['wp'], 'k-', label='all')
ax.plot(wp_sat_result['rpavg'], wp_sat_result['wp'], 'g-', label='satellites')
ax.plot(wp_cen_result['rpavg'], wp_cen_result['wp'], 'b-', label='centrals')
ax.set_xlabel(r'$r_p\ (h^{-1}\mathrm{Mpc})$')
ax.set_ylabel(r'$w_p(r_p)\ (h^{-1}\mathrm{Mpc})$')
ax.legend()
plt.show()
```

```python
wp_red = compute_wp(gpos_red, rp_bins, Lbox, pimax, nthreads)
wp_blue = compute_wp(gpos_blue, rp_bins, Lbox, pimax, nthreads)
```

```python
fig, ax = plt.subplots(figsize=(7,7))

ax.set_xscale('log')
ax.set_yscale('log')
ax.plot(wp_gal_result['rpavg'], wp_gal_result['wp'], 'k-', label='all')
ax.plot(wp_red['rpavg'], wp_red['wp'], 'r-', label='Red galaxies')
ax.plot(wp_blue['rpavg'], wp_blue['wp'], 'b-', label='Blue galaxies')
ax.set_xlabel(r'$r_p\ (h^{-1}\mathrm{Mpc})$')
ax.set_ylabel(r'$w_p(r_p)\ (h^{-1}\mathrm{Mpc})$')
plt.title('2pcf using Corrfunc (Sinha and Garrison)')
ax.legend()
plt.show()
```

# Halo Mass Function:
Shows the fraction of number density that comes due to halos within a small fraction of logarithmic mass slice around M.
Units: $Mpc^{-3}$

```python
Lbox = hr.Lbox
M = halos['m200b']
mass_bins = np.logspace(np.log10(M.min()), np.log10(M.max()), 30)
counts, edges = np.histogram(M, bins=mass_bins)
```

```python
dlnM = np.diff(np.log(edges))
dn_dlnM = (counts / Lbox**3) / dlnM
mass_centers = np.sqrt(edges[1:]*edges[:-1])
```

```python
# since both edges and counts have the same indices
valid = counts > 0
mass_centers = mass_centers[valid]
dn_dlnM = dn_dlnM[valid]

mmin = 40 * hr.mpart
lnMmin = np.log(mmin)
```

```python
fig, ax = plt.subplots(figsize = (6,3))
ax.loglog(mass_centers, dn_dlnM, 'bo',mfc = 'none', ms=4, label='Sahyadri (measured)')
ax.loglog(M_tinker, dndlnM_tinker, '-', color='C1', label='Tinker (2008)')
ax.axvline(mmin, linestyle=':', color = 'black', label='40 particle limit for Sahyadri')
ax.set_xlabel(r'$M_{200b}\ (M_\odot/h_f)$')
ax.set_ylabel(r'$dn/d\ln M\ \ ((h_f/\mathrm{Mpc})^3)$')
ax.set_title('Halo mass function, z=0')
ax.legend()
plt.tight_layout()

ax.set_xlim(left = 1e9)
plt.savefig('halo_mass_function_z0.pdf', dpi=150)
plt.show()
# ax.set_ylim(bottom = 0.4)
```
