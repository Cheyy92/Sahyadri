## HOD Functions:

```python
#///////////////////////////////////////////////////////////////////
# HOD Fits
#///////////////////////////////////////////////////////////////////
class HODFits_BGS(object):
    """ Fitting functions and derivatives specifying best fit 5-parameter HOD fits 
         from mith, Grove, Cole et al. (2024) Public Data Release.
    """
    def __init__(self):
        """ Best fit HOD parameters and derivatives as function of luminosity threshold Mr. """
    
        self._A_min, self._B_min, self._C_min, self._D_min = -0.1163295942035853792, -0.5759066407274642252, 0.1437085418067726994, -0.01908335449679073731
        self._A_sig, self._B_sig, self._C_sig, self._D_sig = 0.01832666435246161490, 0.7584865035087475782, 1.103103765754579246, 0.3277787214124110449
        self._A_0,   self._B_0                    = -0.5468895455660652827, -1.673562380257160864
        self._A_1,   self._B_1, self._C_1, self._D_1         = 1.121497360473139082, -0.4725625165007949491, 0.09200038944813740405, -0.01141762462160326314
        self._A_al,  self._B_al, self._C_al             = 1.111866664500087198, 4.473949866395862784, -4.353309870091147893


    #Fitting functions for HOD Parameters:
    def _M_function(self, Mr, A, B, C, D):
        x = Mr + 20
        return (A + 12) + B*x + C*x**2 + D*x**3          # returns log10(M)

    def _M0_function(self, Mr, A, B):
        x = Mr + 20
        logM0 = (A + 11) + B*x
        return np.maximum(logM0, 1.0)                            # scalar-safe floor

    def _sigma_function(self, Mr, A, B, C, D):
        x = Mr + 20
        return A + (B - A) / (1. + np.exp(C*(x + D)))

    def _alpha_function(self, Mr, A, B, C):
        x = Mr + 20
        return A + B**(-x + C)

    def hod_fit(self, param, Mr):
        """Smith, Grove, Cole et al. (2024) nested BGS HOD, AbacusSummit c000/ph000 best fit.
        Returns Mcut, M1, sigma, alpha, kappa (kappa*Mcut == M0) for simulator.py's HOD_model.
        """
        if param == 'kappa':
            lgMmin = self._M_function(Mr, self._A_min, self._B_min, self._C_min, self._D_min)
            Mcut = 10**lgMmin
            logM0   = self._M0_function(Mr, self._A_0, self._B_0)
            M0   = 10**logM0
            kappa = M0 / Mcut
            return kappa
        elif param == 'M1':
            logM1   = self._M_function(Mr, self._A_1, self._B_1, self._C_1, self._D_1)
            M1   = 10**logM1
            return M1
        elif param == 'siglgM':
            sigma   = self._sigma_function(Mr, self._A_sig, self._B_sig, self._C_sig, self._D_sig)
            return sigma
        elif param == 'alpha':
            alpha   = self._alpha_function(Mr, self._A_al, self._B_al, self._C_al)
            return alpha
        elif param == 'Mcut':
            lgMmin = self._M_function(Mr, self._A_min, self._B_min, self._C_min, self._D_min)
            Mcut = 10**lgMmin
            return Mcut
        elif param == 'M0':
            logM0   = self._M0_function(Mr, self._A_0, self._B_0)
            M0   = 10**logM0
            return M0
        elif param == 'prs':
            # From the Paul et al, 2019 doi:10.1093/mnras/stz1764 # Assuming the SDSS colour assignment works for DESI BGS catalog too
            a0,a1,a2 = 0.773, -0.065, 0.008
            x = Mr + 20.5
            out = (a0 + a1*x + a2*x**2)
            return out

    # def hod_fit_deriv(self,param,Mr):
    #     """ Return derivative wrt Mr for HOD parameter param (string) at luminosity threshold Mr.
    #          param can be one of 'lgMmin','siglgM','lgM1','lgM0','alpha'.
    #     """
    #     a0,a1,a2 = self.bf[param]
    #     x = Mr + 20.5
    #     out = (a1 + 2*a2*x
    #            if param != 'siglgM' else
    #            (a1/a2)*2/np.sqrt(np.pi)*np.exp(-(x/a2)**2))
    #     return out

    

#///////////////////////////////////////////////////////////////////
# HOD Functions
#///////////////////////////////////////////////////////////////////
class HODFunctions(HODFits_BGS):
    """ Functions and derivatives specifying 5-parameter HODs 
         Uses fits from Smith et al, 2024.
    """
    def __init__(self):
        """ HOD functions and derivatives as function of luminosity threshold Mr. """

        HODFits_BGS.__init__(self)
        
    def F_spline(self,x):
        """
        The Cumulative integral function as defined in arXiv:2312.08792v2 
        The integral was analytically calculated using the spline function defined in the paper.
        """
        t = x / np.sqrt(6)
        at = np.abs(t)
        sign = np.sign(x)

        piece1 = (8/3)*t - (16/3)*t**3 + 4*t**4
        piece2 = np.sign(t) * (1 - (4/3)*(1 - at)**4)
        piece3 = np.sign(t) * 1.0

        result = np.where(
            at <= 0.5, piece1,
            np.where(at <= 1, piece2, piece3)
        )
        return result
    
        
    def F_arg(self,Mr,lgm):
        """ Convenience function.
             If both arguments are arrays, output has shape (lgm.size,Mr.size).
        """
        Mr_sc = np.isscalar(Mr)
        lgm_sc = np.isscalar(lgm)

        lgm_arr = np.outer(lgm,np.ones(Mr.size)) if ((not Mr_sc) & (not lgm_sc)) else 1.0*lgm
        arg = (lgm_arr - np.log10(self.hod_fit('Mcut',Mr)))/self.hod_fit('siglgM',Mr)
        return arg

    def erf_arg(self,Mr,lgm):
        """ Convenience function.
             If both arguments are arrays, output has shape (lgm.size,Mr.size).
        """
        Mr_sc = np.isscalar(Mr)
        lgm_sc = np.isscalar(lgm)

        lgm_arr = np.outer(lgm,np.ones(Mr.size)) if ((not Mr_sc) & (not lgm_sc)) else 1.0*lgm
        erfarg = (lgm_arr - np.log10(self.hod_fit('Mcut',Mr)))/self.hod_fit('siglgM',Mr)
        return erfarg

    def fcen_thresh(self,Mr,lgm,hod):
        """ fcen(>L|m) = (1/2)[1 + erf(log10(m/Mmin)/siglgM)]. 
             If both arguments are arrays, output has shape (lgm.size,Mr.size).
        """
        if(hod=='erf'):
            erfarg = self.erf_arg(Mr,lgm)
            out = 0.5*(1+sysp.erf(erfarg))
            return out
        elif(hod=='BGS'):
            Farg = self.F_arg(Mr, lgm)
            out = 0.5*(1+self.F_spline(Farg))
            return out

    def Nsat_thresh(self,Mr,lgm):
        """ Nsat(>L|m) = ((m - M0)/M1)^alpha.
             If both arguments are arrays, output has shape (lgm.size,Mr.size).
        """
        M0 = self.hod_fit('M0',Mr)
        M1 = self.hod_fit('M1',Mr)
        alpha = self.hod_fit('alpha',Mr)
        mhalo = 10**lgm

        Mr_sc = np.isscalar(Mr)
        lgm_sc = np.isscalar(lgm)

        if Mr_sc & lgm_sc:
            out = ((mhalo - M0)/M1)**alpha if mhalo > M0 else 0.0
        else:
            mhalo_arr = np.outer(mhalo,np.ones(Mr.size)) if ((not Mr_sc) & (not lgm_sc)) else 1.0*mhalo
            Dm = (mhalo_arr - M0)/M1
            Dm[Dm < 0.0] = 0.0
            out = Dm**alpha

        return out
    
    def mean_color_sat(self, Mr):
        return 0.83 - 0.08*(Mr + 20)
    
    def mean_color_red(self, Mr):
        return 0.932 - 0.032*(Mr + 20)
    
    def std_color_red(self, Mr):
        return 0.07 + 0.01*(Mr+20)
    
    def mean_color_blue(self, Mr):
        return 0.62 - 0.11*(Mr+20)
    
    def std_color_blue(self, Mr):
        return 0.12 + 0.02*(Mr+20)
    
    def blue_frac(self, Mr):
        return 0.46 + 0.07*(Mr+20)
    
    def p_red_sat(self, Mr):
        num = self.mean_color_sat(Mr) - self.mean_color_blue(Mr)
        den = self.mean_color_red(Mr) - self.mean_color_blue(Mr)
        return num/den
    
```

```python
class Mocker(HODFunctions):
    def __init__(self, real, snap, sim_stem, hod='BGS'):
        self.real = real
        self.snap = snap
        self.sim_stem = sim_stem
        
        print("Initializing SnapshotReader...")
        sr = SnapshotReader(sim_stem=self.sim_stem, real=self.real, snap=self.snap, use_compressed=True)
        
        HODFunctions.__init__(self)
        self.redshift = sr.redshift
        self.Om = sr.Om
        self.Lbox = sr.Lbox # Mpc/h
        self.massdef = 'm200b'
        self.mmin = 2e11 # Msun/h
        self.Mrmax = -18.0 #Maximum(faintest) r-band absolute magnitude for galaxies
        self.Dvir = 200*self.Om #Virial density threshold
        
        self.rhoc = 2.7754e11 # (Msun/h) / (Mpc/h)^3 - Critical Density pf the universe at z=0
        self.rhoc_z = self.rhoc*self.EHub(self.redshift)**2 #Critical density at a particular redshift
        
        self.rng = np.random.RandomState(seed=10) #Random number generator

        dMr = 0.001 #Bin spacing for Mrvals
        nMr = int((23.5+self.Mrmax)/dMr) #number of Mrvals
        self.Mrvals = np.linspace(-23.5,self.Mrmax,nMr) # useful for interpolation later
        self.dMr = self.Mrvals[1]-self.Mrvals[0]
        
        self.hod = hod
        
        # self.halodatatype = {'ID':int,'descID':int,
        #                     'mbnd_vir':float,'vmax':float,'vrms':float,'rvir':float,'rs':float,'np':int,
        #                     'x':float,'y':float,'z':float,'vx':float,'vy':float,'vz':float,
        #                     'Jx':float,'Jy':float,'Jz':float,'spin':float,'rs_klypin':float,
        #                     'mvir':float,'m200b':float,'m200c':float,'mCustom2':float,'mCustom':float,
        #                     'Xoff':float,'Voff':float,'spin_bullock':float,'b_to_a':float,'c_to_a':float,
        #                     'Ax':float,'Ay':float,'Az':float,'b_to_a_500c':float,'c_to_a_500c':float,
        #                     'Ax_500c':float,'Ay_500c':float,'Az_500c':float,
        #                     'TbyU':float,'Mpe_Behroozi':float,'Mpe_Diemer':float,'halfmassradius':float,
        #                     'pid':int} 

        
    ############################################################
    def EHub(self,z):
        """ Computes the dimensionless Hubble Parameter and a given redshift z."""
        return np.sqrt(self.Om*(1+z)**3 + 1-self.Om)
    ############################################################
    
    ############################################################
    # def read_this(self):
    #     halos = pd.read_csv(self.halo_cat,dtype=self.halodatatype,names=self.halodatatype.keys(),
    #                         comment='#',sep=r'\s+',header=None).to_records()
    #     return halos
    ############################################################
    

    ############################################################
    def prep_halo_data(self, massdef='m200b', QE=0.5,Npmin=0, sorthalos=False, keep_subhalos=False):
        """ Reads and cleans catalog by selecting parent objects with mass at least mmin.
            Returns structured array for full halo properties.
        """ 
        hr = HaloReader(sim_stem= self.sim_stem, real=self.real, snap=self.snap)
        hpos, halos = hr.prep_halos(massdef=massdef, QE=QE, Npmin=Npmin, sorthalos=sorthalos, keep_subhalos=keep_subhalos)
        
#         halos = self.read_this()
#         Nhalos_all = halos.size
#         cond_clean = (halos[self.massdef] >= self.mmin)
#         cond_clean = cond_clean & (halos['pid'] == -1)

#         halos = halos[cond_clean]
#         print("Kept {0:d} of {1:d} objects in catalog".format(halos.size,Nhalos_all))

#         del cond_clean
#         gc.collect()

        return hpos, halos
    ############################################################


    ############################################################
    def occupy_halos(self,halo_mass):
        # We use the Monte Carlo Rejection Smpling technique to occupy halos
        """ Decide which halos are occupied. 
             Input is 1-d array of halo masses. Output is 1-d boolean array of same size.
        """
        print("Occupying halos... ")

        occupy = np.ones(halo_mass.size,dtype=bool)
        fcen_min = self.fcen_thresh(self.Mrmax,np.log10(halo_mass), hod=self.hod)
        u = self.rng.rand(halo_mass.size) #draws a uniform random number between 0.0 to 1.0 for every halo(dice)
        #For any given halo, if the randum number u is greater than the calculated probability, it is unoccupied
        occupy[u > fcen_min] = False 

        print_string = "... {0:d} of {1:d} halos occupied\n".format(np.where(occupy)[0].size,occupy.size)
        print_string += "--------------------------------"
        print(print_string)

        return occupy
    ############################################################
    

    ############################################################
    def assign_centrals(self,halos):
        """ Allocate memory and assign (some) central properties. """

        print("Assigning values for centrals... ")
        ####################################################
        # Allocate memory for centrals
        centrals = np.zeros(halos.size,dtype=[('haloid','int'),
                                              ('lgm','double'),('rvir','double'),('con','double'),
                                              ('x','double'),('y','double'),('z','double'),
                                              ('Mr','double'),('Nsat','int'),('color', 'U4')])

        print("... halo IDs, halo masses, positions")
        centrals['haloid'] = halos['ID']
        centrals['lgm'] = np.log10(halos[self.massdef])
        centrals['x'] = halos['x']
        centrals['y'] = halos['y']
        centrals['z'] = halos['z']
        
        rvir = (3*halos[self.massdef]/(4*np.pi*self.Dvir*self.rhoc_z))**(1/3.)
        # Units Mpc/h [physical]
        centrals['rvir'] = rvir*(1+self.redshift) # store comoving value
        centrals['con'] = rvir/(halos['rs']*1e-3/(1+self.redshift)) # rs is comoving kpc/h
        
        ####################################################
        # Assign central luminosity based on mass of halo
        centrals['Mr'] = self.assign_central_luminosity(centrals['lgm'])

        ####################################################
        # Poisson numbers for Nsat: NUmber of satellites for each central
        centrals['Nsat'] = self.get_Poisson(centrals['lgm'])

        print_string = '... done\n'
        print_string += "--------------------------------"
        print(print_string)

        return centrals
    ############################################################
        


    ############################################################
    def get_Poisson(self,lgm_halos):
        """ Calculate mean Nsat and generate Poisson numbers of satellites. 
            Can be generalised to depend on more than halo mass.
        """
        print("... Poisson satellite counts")
        mean_Nsat = self.Nsat_thresh(self.Mrmax,lgm_halos)
        Nsat_values = self.rng.poisson(mean_Nsat)

        return Nsat_values
    ############################################################

    ########################################################
    def assign_satellites(self,centrals):
        """ Allocate memory and assign (some) satellite values."""
        print("Assigning values for satellites... ")

        Nsat_tot = np.sum(centrals['Nsat']) #Total Number of satellites to be assigned

        ####################################################
        # Allocate memory for satellites
        satellites = np.zeros(Nsat_tot,dtype=[('haloid','int'),
                                              ('lgm','double'),('con','double'),
                                              ('x','double'),('y','double'),('z','double'),
                                              ('Mr','double'), ('color', 'U4')])

        print("... halo properties")
        s_lo = 0
        for h in range(centrals.size):
            s_hi = s_lo + centrals['Nsat'][h]
            ################################################
            if centrals['Nsat'][h]:
                sl = np.s_[s_lo:s_hi]
                ############################################
                satellites['haloid'][sl] = centrals['haloid'][h]
                satellites['lgm'][sl] = centrals['lgm'][h]
                satellites['con'][sl] = centrals['con'][h]
            ################################################
            # Update
            s_lo = s_hi

        print("... positions")
        s_lo = 0
        for h in range(centrals.size):
            s_hi = s_lo + centrals['Nsat'][h]
            ################################################
            if centrals['Nsat'][h]:
                sl = np.s_[s_lo:s_hi]
                ############################################
                sx,sy,sz = self.assign_satellite_pos(centrals,h)
                satellites['x'][sl] = sx
                satellites['y'][sl] = sy
                satellites['z'][sl] = sz
            ################################################
            # Update
            s_lo = s_hi

        print("... luminosities")
        s_lo = 0
        for h in range(centrals.size):
            s_hi = s_lo + centrals['Nsat'][h]
            ################################################
            if centrals['Nsat'][h]:
                ############################################
                sMr = self.assign_satellite_luminosities(centrals,h)
                satellites['Mr'][s_lo:s_hi] = sMr
            ################################################
            # Update
            s_lo = s_hi
            
        print("... color")
        satellites['color'] = self.assign_satellite_colors(satellites)
        
        gc.collect
        

        print_string = '... done\n'
        print_string += "--------------------------------"
        print(print_string)

        return satellites
    ########################################################
    
    ########################################################
    def assign_satellite_pos(self,centrals,h):
        """ Assign satellite positions and velocities acc to NFW profile using halo concentration. """
        # Number of satellites in this halo
        Nsat_h = centrals['Nsat'][h]
        # Virial radius [comoving Mpc/h]
        rvir = centrals['rvir'][h]
        con = centrals['con'][h]

        sat_pos = self.gen_NFW_profile(Nsat_h,cvir=con,Rvir=rvir,rng=self.rng)
        sx = sat_pos[:,0] + centrals['x'][h]
        sy = sat_pos[:,1] + centrals['y'][h]
        sz = sat_pos[:,2] + centrals['z'][h]

        sx = sx % self.Lbox
        sy = sy % self.Lbox
        sz = sz % self.Lbox

        return sx,sy,sz
    
    
    ############################################################
    
    ############################################################
    def gen_NFW_profile(self,Nsat,cvir=None,Rvir=None,rng=None,include_cen=False,clean_up=False):
        """ Generate 3-d points distributed according to specified radial profile.
            Returns array of shape (Ngal,3) where Ngal = Nsat if include_cen = False else Nsat+1
            and central is assumed to be at origin. (Can include spatial offset later.)
        """
        if rng is None:
            rng = np.random.RandomState(seed=10)
        phi = 2*np.pi*rng.rand(Nsat)
        cos_theta = 2*rng.rand(Nsat) - 1
        sin_theta = np.sqrt(1-cos_theta**2)
        
        rsamp = self.gen_rsamp(Nsat,cvir=cvir,Rvir=Rvir,rng=rng,clean_up=clean_up)

        x_trc = rsamp*sin_theta*np.sin(phi)
        y_trc = rsamp*sin_theta*np.cos(phi)
        z_trc = rsamp*cos_theta

        sample = np.array([x_trc,y_trc,z_trc]).T
        if include_cen:
            sample = np.append(sample,[[0.,0.,0.]],axis=0)
            
        if clean_up:
            del x_trc,y_trc,z_trc
            gc.collect()

        return sample
    ############################################################
    
    ############################################################
    def gen_rsamp(self,Nsat,cvir=None,Rvir=None,rng=None,clean_up=False):
        """ Generate radial location sample for speficied profile. 
            Return array of r values of shape (Nsat,).
        """
        xmax = 2*cvir

        if rng is None:
            rng = np.random.RandomState(seed = 10)

        xfine = np.linspace(0,xmax,100000)
        dx = xfine[1]-xfine[0]
        Px = np.zeros_like(xfine)
        
        rs = Rvir/cvir
        Px = self.Mencl_nfw(xfine)
        ind_small = np.where(xfine < 1e-3)[0]
        if ind_small.size:
            Px[ind_small] = xfine[ind_small]**2/2 - 2*xfine[ind_small]**3/3.0 
            Px[ind_small] = Px[ind_small] + 3*xfine[ind_small]**4/4.0 - 4*xfine[ind_small]**5/5.0
        Px /= self.Mencl_nfw(cvir)
                
        vran = rng.rand(Nsat)
        rsamp = np.interp(vran,Px,xfine)
        if np.any(rsamp <= dx):
            print('! Warning ! : Ignore separations x < {0:.2e}'.format(dx))

        if clean_up:
            del xfine,Px
            gc.collect()
        
        rsamp *= rs
        
        return rsamp
    ############################################################
    
    ############################################################
    def Mencl_nfw(self,x):
        """ Convenience function for NFW enclosed mass. Expect x = r/rs. """
        return np.log(1+x) - x/(1+x)
    ############################################################
    
    
    
    ########################################################
    def assign_central_luminosity(self,lgm_halos):
        """ Assign central luminosities. Can be generalised to depend on more than halo mass. """

        print("... luminosities")
        n = lgm_halos.size
        Mr_cen = np.empty(n, dtype=float)
        u = self.rng.rand(n)
        Mrvals = self.Mrvals
        nmax = Mrvals.size - 1

        for h in range(n):
            PcenL = self.fcen_thresh(Mrvals, lgm_halos[h], self.hod) / self.fcen_thresh(self.Mrmax, lgm_halos[h], self.hod)
            idx = np.searchsorted(PcenL, u[h])
            Mr_cen[h] = Mrvals[idx if idx < nmax else nmax]

        print('done!')
        del u
        gc.collect()
        return Mr_cen
    ########################################################
    
    ########################################################
    def assign_satellite_luminosities(self,centrals,h):
        """ Assign satellite luminosities. """

        Nsat_h = centrals['Nsat'][h]
        Mr_cen = centrals['Mr'][h]
        lgmhalo = centrals['lgm'][h]
        PsatL = self.Nsat_thresh(self.Mrvals,lgmhalo)/self.Nsat_thresh(self.Mrmax,lgmhalo)

        ####################################################
        # Generate luminosities, without worrying about Lcen.
        # ... generate Nsat_h uniform deviates
        u = self.rng.rand(Nsat_h)
        # ... find largest L (smallest Mr) where Psat(>L|m) > u
        Mr_sat = np.array([self.Mrvals[(PsatL >= u[s])][0] for s in range(Nsat_h)])
        return Mr_sat
    ########################################################
    
    def assign_centrals_colors(self, centrals, satellites, Mr_bin_edges = [-23, -22, -21, -20, -19, -18]):
        color = np.empty(centrals.size, dtype ='U4')
        u = self.rng.rand(centrals.size)
        
        #From Zehavi et al(SDSS), 2011
        p_red_bin = {
            (-23, -22): 0.82,
            (-22, -21): 0.63,
            (-21, -20): 0.53,
            (-20, -19): 0.43,
            (-19, -18): 0.28
        }
        
        for i in range(len(Mr_bin_edges)-1):
            Mr_hi, Mr_lo = Mr_bin_edges[i], Mr_bin_edges[i+1]
            bin_key = (Mr_hi, Mr_lo)
            p_red = p_red_bin.get(bin_key)
            
            #Fractions:
            in_bin_cen = (centrals['Mr'] >= Mr_hi) & (centrals['Mr'] < Mr_lo)
            in_bin_sat = (satellites['Mr'] >= Mr_hi) & (satellites['Mr'] < Mr_lo)
            n_cen, n_sat = in_bin_cen.sum(), in_bin_sat.sum()
            p_cen_bar = n_cen/(n_cen+n_sat)
            p_sat_bar = n_sat/(n_cen+n_sat)
            
            #prs:
            prs_per_gal = np.clip(self.hod_fit('prs', centrals['Mr'][in_bin_cen]), 0.01, 0.99)
            
            #Eq A.10 from paul,pahwa,paranjape(2019):
            p_red_cen  =(p_red - prs_per_gal*p_sat_bar)/p_cen_bar
            is_red = u[in_bin_cen] < p_red_cen
            color[in_bin_cen] = np.where(is_red, 'red', 'blue')
            
        return color
    
    ########################################################
    def assign_satellite_colors(self, satellites):
        """ Red/blue assignment for satellites using prs(Mr), indipendent of halo mass"""
        
        prs = self.hod_fit('prs', satellites['Mr'])
        prs = np.clip(prs, 0.01, 0.99)
        u = self.rng.rand(satellites.size)
        is_red = u < prs
        color = np.where(is_red, 'red', 'blue')

        return color
    ########################################################
    
    def assign_satellite_gr(self, satellites, add_scatter=False):
        u = self.rng.uniform(0,1, satellites.shape[0])
        is_red = u < self.p_red_sat(satellites['Mr'])
        gr_color = np.zeros(satellites['Mr'].shape[0])
        if np.any(is_red):
            mr_red = satellites['Mr'][is_red]
            mean_red = self.mean_color_red(satellites['Mr'][is_red])
            if add_scatter:
                sigma_red = self.std_color_red(mr_red) # Assuming a scatter function exists
                mean_red += self.rng.normal(0, sigma_red)    
            gr_color[is_red] = mean_red
        
        if np.any(~is_red):
            mr_blue = satellites['Mr'][~is_red]
            mean_blue = self.mean_color_red(satellites['Mr'][~is_red])
            if add_scatter:
                sigma_blue = self.std_color_blue(satellites['Mr'][~is_red])
                mean_blue += self.rng.normal(0, sigma_blue)
            gr_color[~is_red] = mean_blue
        
        return gr_color
        
    
    ########################################################
    
    def assign_central_gr(self, centrals, nsat, add_scatter=False):
        
        u = self.rng.uniform(0,1, centrals.shape[0])
        ncen = centrals['Mr'].shape[0]
        cen_frac = 1/(1+nsat/ncen)
        is_red = u > self.blue_frac(centrals['Mr'])/cen_frac
        gr_color = np.zeros(ncen)
        
        if np.any(is_red):
            mr_red = centrals['Mr'][is_red]
            mean_red = self.mean_color_red(centrals['Mr'][is_red])
            if add_scatter:
                sigma_red = self.std_color_red(mr_red) # Assuming a scatter function exists
                mean_red += self.rng.normal(0, sigma_red)    
            gr_color[is_red] = mean_red
          
        if np.any(~is_red):
            mr_blue = centrals['Mr'][~is_red]
            mean_blue = self.mean_color_red(centrals['Mr'][~is_red])
            if add_scatter:
                sigma_blue = self.std_color_blue(centrals['Mr'][~is_red])
                mean_blue += self.rng.normal(0, sigma_blue)
            gr_color[~is_red] = mean_blue
        
        return gr_color
    ############################################################
    
    ############################################################
    def mock_it(self, massdef='m200b', QE=0.5,Npmin=0, sorthalos=False, keep_subhalos=False):
        """ Main execution routine. """        
        # halo data
        
        hr = HaloReader(sim_stem= self.sim_stem, real=self.real, snap=self.snap)
        print('HaloReader Initialized.')
        print('Reading halos...')
        hpos, halos_all = hr.prep_halos(massdef=massdef, QE=QE, Npmin=Npmin, sorthalos=sorthalos, keep_subhalos=keep_subhalos)
        # hpos, halos_all = self.prep_halo_data()

        # occupy halos
        occupy = self.occupy_halos(halos_all[self.massdef])
        halos = halos_all[occupy]
        all_masses = halos_all[self.massdef]
        del halos_all
        gc.collect()

        # galaxy properties except color and derived props
        centrals = self.assign_centrals(halos)
        satellites = self.assign_satellites(centrals)
        
        print("...Assigning centrals color")
        centrals['color'] = self.assign_centrals_colors(centrals=centrals, satellites=satellites)
        
        del halos
        gc.collect()

        print_string = "... {0:d} galaxies created\n".format(centrals.size + satellites.size)
        print_string += "--------------------------------"
        print(print_string)

        print_string = '... all done\n'
        print_string += "--------------------------------"
        print(print_string)

        return centrals,satellites,all_masses

    ############################################################
```
