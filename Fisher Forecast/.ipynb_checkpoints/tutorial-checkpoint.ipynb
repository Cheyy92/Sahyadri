```python
## On how to build covariance and fisher forecast.
import numpy as np  #numpy: https://numpy.org/doc/stable/index.html
import importlib.util  #importlib: https://docs.python.org/3/library/importlib.html
from scipy.spatial import cKDTree
from astropy.table import Table
from numpy.lib import recfunctions as rfn
import matplotlib
import matplotlib.pyplot as plt
import time

import os, sys
root = "/mnt/home/project/csatsangi.chetan/projects"
sys.path.insert(0, os.path.join(root, "sahyadri-sandbox", "scripts", "post-process"))
```

```python
jackknife_vvf_100 = Table.read(root + '/JackKnife/rsd_results/vvf_results_100.fits').as_array()
```

```python
vvf_realizations = jackknife_vvf_100['vvf_realizations']
njn = vvf_realizations.shape[1]

# 1. Calculate the mean across the jackknife regions (shape: n_percentiles)
vvf_mean_jn = np.mean(vvf_realizations, axis=1)

# 2. Calculate the deviations from the mean (shape: n_percentiles, njn)
# The [:, None] broadcasts the 1D mean array across the njn columns
delta = vvf_realizations - vvf_mean_jn[:, None]

# 3. Compute the Jackknife covariance matrix (shape: n_percentiles, n_percentiles)
cov_matrix = ((njn - 1) / njn) * (delta @ delta.T)

# 4. Invert to get the precision matrix
C_inv = np.linalg.inv(cov_matrix)
```

$$\mathbf{F} = \underbrace{\mathbf{D}^T}_{(m+n) \times p} \times \underbrace{\mathbf{C}^{-1}}_{p \times p} \times \underbrace{\mathbf{D}}_{p \times (m+n)}$$

```python
def read_deriv_vvf_2_5(param_name, rsd_suff):
    """
    param_name can take: alpha, kappa, logM1, logMcut, sigma
    """
    tab = Table.read(f'{root}/Derivatives/derivatives/2_5/dVVF_d{param_name}_{rsd_suff}.fits')
    
    # Extract the raw NumPy array for that specific column
    d_vvf_d_param = tab[f'dVVF_d{param_name}'].data
    return d_vvf_d_param
```

```python
# Example 1D derivative arrays of length p (representing d(VVF)/d(theta))

d_vvf_d_logM1 = read_deriv_vvf_2_5('logM1', 'rsd')
d_vvf_d_sigma = read_deriv_vvf_2_5('sigma', 'rsd')
d_vvf_d_logMcut = read_deriv_vvf_2_5('logMcut', 'rsd')
d_vvf_d_alpha = read_deriv_vvf_2_5('alpha', 'rsd')
d_vvf_d_kappa = read_deriv_vvf_2_5('kappa', 'rsd')

# Stack them column-wise to form the Jacobian D
# Shape of D will be (p, 5) since m = 5
D = np.column_stack((d_vvf_d_alpha, d_vvf_d_kappa, d_vvf_d_sigma, d_vvf_d_logM1, d_vvf_d_logMcut))
```

```python
D.shape
```

```python
# D.T is (3, p)
# C_inv is (p, p)
# D is (p, 3)

# The @ operator performs matrix multiplication in NumPy
F = D.T @ C_inv @ D

# The resulting Fisher matrix F has shape (3, 3)
```

```python
# Invert the Fisher matrix (shape: 3, 3)
F_inv = np.linalg.inv(F)

# Extract the 1-sigma errors for each HOD parameter
# np.diag extracts the diagonal elements of the matrix
errors = np.sqrt(np.diag(F_inv))

sigma_Mmin = errors[0]
sigma_sigma = errors[1]
sigma_M0 = errors[2]

print(f"1-sigma error on log M_min: {sigma_Mmin}")
print(f"1-sigma error on sigma_logM: {sigma_sigma}")
print(f"1-sigma error on log M_0: {sigma_M0}")
```
